# NumPy and shapes

Most of the arrays used in tabular machine learning have a simple interpretation: each row is an example and each column is a feature. Keeping that interpretation consistent is more important than memorizing individual NumPy operations. The shapes below describe the linear predictor used throughout the later notebooks.


## The data and parameter shapes

For `n` examples and `d` features, I use the following convention:

```text
X       (n, d)   feature matrix
w       (d,)     one weight per feature
b       scalar   one shared bias
y       (n,)     one target per example
X @ w   (n,)     one prediction per example
```

The first code cell makes these dimensions explicit before any arithmetic is performed.


````python
import numpy as np

X = np.array([[1., 2.],
              [3., 4.],
              [5., 6.]])
w = np.array([0.5, -1.0])
b = 0.25
y = np.array([0., 1., 1.])

print('X:', X.shape)
print('w:', w.shape)
print('b:', np.shape(b))
print('y:', y.shape)

assert X.shape == (3, 2)
assert w.shape == (2,)
assert y.shape == (3,)
````

## A prediction and its residual

`X @ w` computes the dot product between every row of `X` and the weight vector. Adding the scalar `b` shifts every prediction by the same amount. Once the prediction has the same shape as `y`, the residual and mean squared error are defined element by element.


````python
predictions = X @ w + b
residuals = predictions - y
mse = np.mean(residuals ** 2)

print('X @ w:', (X @ w).shape)
print('predictions:', predictions)
print('residuals:', residuals)
print('MSE:', mse)

assert predictions.shape == y.shape
````

## Broadcasting

Broadcasting is useful because it lets a scalar or a row-shaped array participate in an operation without manually copying it. It can also hide a mistake. A vector with shape `(d,)` is broadcast across feature columns, while an array with shape `(n, 1)` is broadcast down rows. Both operations are valid, but they describe different assumptions about which values are shared.


````python
feature_offsets = np.array([10., 20.])
row_offsets = np.array([[100.], [200.], [300.]])

print('feature offsets:\n', X + feature_offsets)
print('row offsets:\n', X + row_offsets)
````

## One-dimensional vectors and column vectors

NumPy treats `(d,)` as a one-dimensional vector rather than as either a row or a column. Reshaping it to `(d, 1)` changes the shape of the result of matrix multiplication. This distinction becomes important when a target is stored as `(n, 1)` but a model returns `(n,)`; subtraction may otherwise broadcast into an `(n, n)` array.


````python
w_column = w.reshape(2, 1)

print('w:', w.shape, 'X @ w:', (X @ w).shape)
print('w_column:', w_column.shape, 'X @ w_column:', (X @ w_column).shape)

assert (X @ w).shape == (3,)
assert (X @ w_column).shape == (3, 1)
````

## Shape checks

The assertions in the final cell are small but useful. They state the assumptions at the boundary between the data, the model, and the loss. A shape error found there is much easier to diagnose than a plausible-looking loss produced several operations later.
